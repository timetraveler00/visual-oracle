# NITS-IQA analysis (English publication adaptation)

This notebook is adapted from a historical experiment notebook for an English-only repository. The original notebook remains in the local archive. Source data are the 405 recorded pairs; no source images or model inference are included. Numeric calculations retain the historical logic, with an equivalent binary balanced-accuracy implementation and explicit repairs of invalid or out-of-order cells.

```python
from pathlib import Path
import pandas as pd

def locate_nits_csv():
    """Find the English-header NITS-IQA scores relative to the notebook or repository."""
    base = Path.cwd().resolve()
    candidates = [base / "nits_iqa/verified/NITS_IQA_405_verified.csv",
                  base / "../verified/NITS_IQA_405_verified.csv",
                  base / "verified/NITS_IQA_405_verified.csv"]
    for candidate in candidates:
        if candidate.is_file():
            return candidate.resolve()
    raise FileNotFoundError("NITS_IQA_405_verified.csv not found; run from repository root or notebook directory")

DATA_CSV = locate_nits_csv()

def fast_balanced_accuracy(y_true, y_pred):
    """Binary balanced accuracy, equivalent to sklearn on these two-class folds."""
    import numpy as np
    yt = np.asarray(y_true)
    yp = np.asarray(y_pred)
    pos = yt == 1
    neg = yt == 0
    if not pos.any() or not neg.any():
        raise ValueError("Both classes are required for this binary benchmark")
    return 0.5 * ((yp[pos] == 1).mean() + (yp[neg] == 0).mean())
```

```python
import numpy as np
import pandas as pd
from pathlib import Path

from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import (
    balanced_accuracy_score,
    accuracy_score,
    f1_score,
    confusion_matrix
)

file = DATA_CSV
df = pd.read_csv(file)

metrics = [
    "DINOv2-Base (%)",
    "DINOv2-Large (%)",
    "Dists (%)",
    "SSIM (%)",
    "OpenCLIP (%)"
]

y = (df["MOS"] >= 50).astype(int).to_numpy()
groups = df["Ref"].to_numpy()

logo = LeaveOneGroupOut()

# Threshold shifts in percentage points
# shifts = [-10, -5, 0, 5, 10]
shifts = [-10, -5, -2, -1, 0, 1, 2, 5, 10]

results = []
threshold_records = []

for metric in metrics:

    scores = df[metric].to_numpy()

    # Separate predictions for each threshold shift
    predictions = {
        shift: np.zeros(len(df), dtype=int)
        for shift in shifts
    }

    for train_idx, test_idx in logo.split(scores, y, groups):

        train_scores = scores[train_idx]
        train_labels = y[train_idx]

        # Select the baseline threshold on the training fold only
        thresholds = np.unique(train_scores)

        best_threshold = thresholds[0]
        best_score = -1

        for threshold in thresholds:

            pred = (train_scores >= threshold).astype(int)

            score = fast_balanced_accuracy(
                train_labels, pred
            )

            if score > best_score:
                best_score = score
                best_threshold = threshold

        threshold_records.append({
            "Metric": metric,
            "Test_Reference": groups[test_idx][0],
            "Threshold": best_threshold
        })

        # Apply alternative shifts to the threshold learned on training data
        for shift in shifts:

            adjusted_threshold = np.clip(
                best_threshold + shift, 0, 100
            )

            predictions[shift][test_idx] = (
                scores[test_idx] >= adjusted_threshold
            ).astype(int)

    # Evaluate each threshold shift
    for shift in shifts:

        pred = predictions[shift]

        tn, fp, fn, tp = confusion_matrix(
            y, pred, labels=[0, 1]
        ).ravel()

        results.append({
            "Metric": metric,
            "Threshold_Shift": shift,
            "Accuracy": accuracy_score(y, pred),
            "F1": f1_score(y, pred),
            "FPR": fp / (fp + tn),
            "FNR": fn / (fn + tp)
        })

sensitivity_df = pd.DataFrame(results)
threshold_df = pd.DataFrame(threshold_records)

# Save results
output_dir = Path("results")
output_dir.mkdir(parents=True, exist_ok=True)

sensitivity_df.to_csv(
    output_dir / "threshold_sensitivity.csv",
    index=False
)

threshold_df.to_csv(
    output_dir / "calibrated_thresholds.csv",
    index=False
)

print("THRESHOLD SENSITIVITY RESULTS")
print(sensitivity_df.round(4).to_string(index=False))

print("\nCALIBRATED THRESHOLDS")
print(threshold_df.to_string(index=False))
```
